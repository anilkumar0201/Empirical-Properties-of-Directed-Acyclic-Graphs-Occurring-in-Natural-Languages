```python
import pandas as pd
import matplotlib.pyplot as plt

# Read file
df = pd.read_excel("Random_structures_english_clean.xlsx", sheet_name="Random_structures_english")

# Frequency tables normalized
real = df[df["type"]=="real"]["max_arity"].value_counts(normalize=True).sort_index()
rand = df[df["type"]=="random"]["max_arity"].value_counts(normalize=True).sort_index()

# Combine indexes
idx = sorted(set(real.index).union(set(rand.index)))

real = real.reindex(idx, fill_value=0)
rand = rand.reindex(idx, fill_value=0)

# Plot
x = range(len(idx))
w = 0.4

plt.figure(figsize=(10,6))
plt.bar([i-w/2 for i in x], real.values, width=w, label="Real")
plt.bar([i+w/2 for i in x], rand.values, width=w, label="Random")

plt.xticks(x, idx)
plt.xlabel("max_arity")
plt.ylabel("Proportion")
plt.title("Normalized Distribution of max_arity for english")
plt.legend()
plt.savefig("Normalized Distribution of max_arity for english.png", dpi=300, bbox_inches='tight')   # save first
plt.show()                                                          # show after saving

from google.colab import files
files.download("Normalized Distribution of max_arity for english.png")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from scipy.interpolate import make_interp_spline

# Load data
df = pd.read_excel("Random_structures_english_clean.xlsx",
                   sheet_name="Random_structures_english")

# Fix column name
df = df.rename(columns={'n/depth': 'density'})

# Separate real and random
real_df = df[df['type'] == 'real']
random_df = df[df['type'] == 'random']


# Smooth curve function
def smooth_curve(x, y):
    x = np.array(x)
    y = np.array(y)

    # Sort values
    sorted_idx = np.argsort(x)
    x = x[sorted_idx]
    y = y[sorted_idx]

    # Smooth x-axis
    x_new = np.linspace(x.min(), x.max(), 300)

    # Handle small data
    k = 3 if len(x) > 3 else 2

    spline = make_interp_spline(x, y, k=k)
    y_smooth = spline(x_new)

    return x_new, y_smooth


# Plot + Save function
def plot_distribution(x_col, y_col, title, ylabel, filename):
    plt.figure(figsize=(8,5))

    # Group by n
    real_group = real_df.groupby(x_col)[y_col].mean()
    random_group = random_df.groupby(x_col)[y_col].mean()

    # Smooth curves
    x_real, y_real = smooth_curve(real_group.index, real_group.values)
    x_rand, y_rand = smooth_curve(random_group.index, random_group.values)

    # Plot lines
    plt.plot(x_real, y_real, label='Real')
    plt.plot(x_rand, y_rand, label='Random')

    # Scatter points
    plt.scatter(real_group.index, real_group.values, s=20)
    plt.scatter(random_group.index, random_group.values, s=20)

    plt.xlabel('n')
    plt.ylabel(ylabel)
    plt.title(title)
    plt.legend()

    # Save PNG
    plt.savefig(filename, dpi=300, bbox_inches='tight')

    plt.show()


# Generate all plots
plot_distribution('n', 'projection_degree',
                  'Projection Degree vs n [English]',
                  'Projection Degree',
                  'projection_degree.png')

plot_distribution('n', 'density',
                  'Density vs n [English]',
                  'Density',
                  'density.png')

plot_distribution('n', 'max_arity',
                  'Max Arity vs n [English]',
                  'Max Arity',
                  'max_arity.png')

plot_distribution('n', 'avg_arity',
                  'Avg Arity vs n [English]',
                  'Average Arity',
                  'avg_arity.png')
```
